```python
from yellowbrick.classifier import ConfusionMatrix
```

```python
cm_credit = ConfusionMatrix(naive_credit_data)
cm_credit.fit(X_credit_trainamento, Y_credit_trainamento)
cm_credit.score(X_credit_teste, Y_credit_teste)
```
